```python
# Safely drop 'Unnamed: 0' only if it exists in the DataFrame
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])
    print("Dropped 'Unnamed: 0' successfully.")
else:
    print("Column 'Unnamed: 0' is not present in this dataset. Skipping...")

print(df.head())
```

```python
import matplotlib.pyplot as plt
```

```python
# changing datatypes for the deprivation dataset
df['date'] = pd.to_datetime(df['date'])
df['year'] = pd.to_numeric(df['year'], errors='coerce').astype('Int64')
df['date'].head()
```

```python
# adding new columns
df['gender_gap_pct'] = df['women_pct'] - df['men_pct']
df['age_group_average_pct'] = df[['children_pct', 'adults_18_64_pct', 'seniors_65_plus_pct']].mean(axis=1)
print('New columns added successfully!')
print(df[['country', 'year', 'gender_gap_pct', 'age_group_average_pct']].head())
```

```python
# checking null values
df.isnull().sum()
```

```python
# filling missing numeric values using column means
numeric_columns = ['deprivation_rate_pct', 'women_pct', 'men_pct', 'children_pct', 'adults_18_64_pct', 'seniors_65_plus_pct', 'gender_gap_pct', 'age_group_average_pct']
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())
```

```python
# checking null values after filling
df.isnull().sum()
```

```python
df['double_deprivation_rate_pct'] = df['deprivation_rate_pct'].apply(lambda x: x * 2)
print(df[['deprivation_rate_pct', 'double_deprivation_rate_pct']].head())
```

```python
# grouping country and finding the average deprivation rate
country_deprivation = df.groupby('country')['deprivation_rate_pct'].mean().sort_values(ascending=False)
print(country_deprivation.head(10))
```

```python
# grouping year and finding the average deprivation rate
year_deprivation = df.groupby('year')['deprivation_rate_pct'].mean()
print(year_deprivation)
```

```python
# grouping country and finding the average gender gap
country_gender_gap = df.groupby('country')['gender_gap_pct'].mean().sort_values(ascending=False)
print(country_gender_gap.head(10))
```

```python
# counting records for each country
country_counts = df.groupby('country')['country_code'].count().sort_values(ascending=False)
print(country_counts.head(10))
```

```python
# grouping using year and finding average rates for age groups
age_analysis = df.groupby('year')[['children_pct', 'adults_18_64_pct', 'seniors_65_plus_pct']].mean()
print(age_analysis)
```

```python
# sorting countries according to deprivation rate in descending order
high_deprivation = df.sort_values('deprivation_rate_pct', ascending=False)
print(high_deprivation[['country', 'year', 'deprivation_rate_pct']].head(10))
```

```python
# sorting records according to the gender gap
high_gender_gap = df.sort_values('gender_gap_pct', ascending=False)
print(high_gender_gap[['country', 'year', 'gender_gap_pct']].head(10))
```

```python
# counting records for each country
country_counts = df.groupby('country')['country_code'].count()
top_countries = country_counts.sort_values(ascending=False).head(10)
print(top_countries)
```

```python
# counting records for each year
year_counts = df.groupby('year')['country_code'].count().sort_values(ascending=False)
print(year_counts)
```

```python
# bar graph for average deprivation rate by year
year_deprivation.plot(kind='bar')
plt.xlabel('Year', fontsize=12)
plt.ylabel('Average Deprivation Rate (%)', fontsize=12)
plt.title('Average Material Deprivation Rate by Year', fontsize=14, fontweight='bold')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# bar graph for top 10 countries by record count
top_countries.plot(kind='bar')
plt.xlabel('Country', fontsize=12)
plt.ylabel('Number of Records', fontsize=12)
plt.title('Top 10 Countries by Number of Records', fontsize=14, fontweight='bold')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# scatter plot for deprivation rate vs gender gap
plt.scatter(df['gender_gap_pct'], df['deprivation_rate_pct'])
plt.xlabel('Women - Men Rate Difference (%)')
plt.ylabel('Deprivation Rate (%)')
plt.title('Gender Gap vs Material Deprivation Rate')
plt.show()
```

```python
# scatter plot for year vs deprivation rate
plt.scatter(df['year'], df['deprivation_rate_pct'])
plt.xlabel('Year')
plt.ylabel('Deprivation Rate (%)')
plt.title('Year vs Material Deprivation Rate')
plt.show()
```

```python
# grouping year and finding average deprivation rate
print(year_deprivation)
```

```python
# histogram for the distribution of deprivation rates
plt.hist(df['deprivation_rate_pct'], bins=10)
plt.xlabel('Deprivation Rate (%)')
plt.ylabel('Number of Records')
plt.title('Distribution of Material Deprivation Rates')
plt.show()
```

```python
# histogram for the distribution of gender gaps
plt.hist(df['gender_gap_pct'], bins=10)
plt.xlabel('Gender Gap (%)')
plt.ylabel('Number of Records')
plt.title('Distribution of Gender Gaps')
plt.show()
```

```python
# line graph for the average deprivation rate over time
plt.plot(year_deprivation.index, year_deprivation.values, marker='o')
plt.xlabel('Year')
plt.ylabel('Average Deprivation Rate (%)')
plt.title('Average Deprivation Rate Over Time')
plt.grid(True)
plt.show()
```

```python
# pie chart for the latest year's top countries by record count
latest_year = df['year'].max()
latest_counts = df.loc[df['year'] == latest_year, 'country'].value_counts().head(10)
plt.pie(latest_counts, labels=latest_counts.index, autopct='%1.1f%%')
plt.title(f'Top Countries in {latest_year}')
plt.show()
```
